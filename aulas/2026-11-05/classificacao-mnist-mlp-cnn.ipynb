```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense
```

```python
(X_train_raw, y_train), (X_test_raw, y_test) = tf.keras.datasets.mnist.load_data()

# Normalização da escala dos pixels das imagens
X_train = X_train_raw.astype('float32') / 255
X_test = X_test_raw.astype('float32') / 255

# Conversão do canal de cor para escala de cinza
X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)
```

```python
plt.figure(figsize=(10, 4))
for i in range(12):
  plt.subplot(2,6,i+1)
  plt.imshow(X_train_raw[i], cmap='gray')
  plt.title(f'Digito: {y_train[i]}', fontsize=11, fontweight='bold')
  plt.axis('off')
plt.suptitle('Amostras da Base Oficial')
plt.tight_layout()
plt.show()
```

```python
tf.keras.utils.set_random_seed(42)

modelo_denso = Sequential([
  Input(shape=(28, 28, 1)),
  Flatten(name='achata_pixel'),
  Dense(128, activation='relu', name='oculta_densa'),
  Dense(10, activation='softmax', name='saida')
])

modelo_denso.compile(
  optimizer='adam',
  loss='sparse_categorical_crossentropy',
  metrics=['accuracy']
)

modelo_denso.summary()
```

```python
tf.keras.utils.set_random_seed(42)

modelo_cnn = Sequential([
    Input(shape=(28, 28, 1)),

    # Bloco de Filtros: 32 filtros de 3x3 procurando traços
    Conv2D(32, kernel_size=(3, 3), activation='relu', name='conv_1'),

    # Encolher mais ainda a matriz final com MaxPolling
    MaxPooling2D((2, 2), name='pool_1'),

    Flatten(name='achatamento'),
    Dense(64, activation='relu', name='densa'),
    Dense(10, activation='softmax', name='saida')
])

modelo_cnn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

modelo_cnn.summary()
```

```python
historico_cnn = modelo_cnn.fit(X_train,
                               y_train,
                               epochs=5,
                               batch_size=64,
                               validation_split=0.15,
                               verbose=1
                               )

# Avaliação do conjunto de teste cego

loss_teste, acc_teste = modelo_cnn.evaluate(X_test, y_test, verbose=0)
print(f'\nAcurácia no conjunto de teste: {acc_teste * 100:.2f}%')
```

```python
# Classificação correta do número
indice_teste = 1000

imagem_exemplo = X_test[indice_teste]
digito_verdadeiro = y_test[indice_teste]

# Predição para imagem isolada
amostra_batch = np.expand_dims(imagem_exemplo, axis=0)
probs = modelo_cnn.predict(amostra_batch, verbose=0)[0]
predicao_campea = np.argmax(probs)
certeza = probs[predicao_campea] * 100

# Plotagem de imagem e gráfico
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16,4))

# Lado Esquerdo > Imagem Real
ax1.imshow(imagem_exemplo.squeeze(), cmap='gray')
ax1.set_title(f'Dígito Real: {digito_verdadeiro}\nDígito Predito: {predicao_campea}\n(Certeza: {certeza:.1f}%)',
              fontsize=11,
              fontweight='bold')
ax1.axis('off')

# Lado Direito > Imagem Predita
barras = ax2.bar(range(10), probs*100, color = '#94a3b8')
barras[predicao_campea].set_color('#10b981')
ax2.set_xticks(range(10))
ax2.set_xlabel('Classes possíveis (0 a 9):')
ax2.set_ylabel('Certeza da Rede (%)')
ax2.set_ylim(0,105)
ax2.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout
plt.show()
```
