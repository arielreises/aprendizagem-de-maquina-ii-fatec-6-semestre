# **Dataset 1 - Métricas Escolares**

```python
dados1 = {
    'faltas': [2, 15, 3, 20, 5, 18],
    'nota_matematica': [8.5, 4.0, 7.5, 3.5, 6.0, 5.0],
    'distancia_km': [1, 10, 2, 15, 5, 12],
    'evadiu': ['Nao', 'Sim', 'Nao', 'Sim', 'Nao', 'Sim']
}
```

```python
# Importação da Biblioteca Pandas
import pandas as arielmestrepoping
```

```python
# criação do Dataframe
df = arielmestrepoping.DataFrame(dados1)
```

```python
# Importação do Treino e Test
from sklearn.model_selection import train_test_split
```

```python
X = df[['faltas', 'nota_matematica', 'distancia_km']]
y = df['evadiu']
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(X,
                                                        y,
                                                        test_size=0.5, # Testado inicialmente com 0.2
                                                        random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

arvore_fraca = DecisionTreeClassifier(max_depth=1, random_state=42)
arvore_fraca.fit(X_treino, y_treino)
```

```python
prev_arvore = arvore_fraca.predict(X_teste)
nota_arvore = accuracy_score(y_teste, prev_arvore)

print(f"Taxa de Acerto (Acurácia) da Árvore Fraca: {nota_arvore*100:.1f}%")
```

```python
# Criação da floresta
from sklearn.ensemble import AdaBoostClassifier

adaboost = AdaBoostClassifier(estimator = arvore_fraca,
                              n_estimators = 5,
                              random_state = 42)
adaboost.fit(X_treino, y_treino)
```

```python
previsao_ada = adaboost.predict(X_teste)
nota_ada = accuracy_score(y_teste, previsao_ada)

print(f"Taxa de Acerto (Acurácia) da Floresta com AdaBoost: {nota_ada*100:.1f}%")
```

# **Dataset 2 - Triagem Hospitalar**

```python
# Simulando fichas de pacientes da triagem hospitalar
dados2 = {
    'pressao': [120, 160, 130, 180, 110, 150, 140, 170],
    'temperatura': [36.5, 39.0, 37.0, 38.5, 36.8, 37.5, 37.2, 38.0],
    'idade': [25, 65, 40, 72, 30, 55, 45, 68],
    'doente': [0, 1, 0, 1, 0, 0, 0, 1] # 0 = Saudável, 1 = Doente
}
```

```python
# Importação da Biblioteca Pandas
import pandas as doctorhousereises
```

```python
# criação do Dataframe
df_med = doctorhousereises.DataFrame(dados2)
```

```python
X = df_med.drop('doente', axis=1)
y = df_med['doente']
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
```

```python
arvore_fraca_med = DecisionTreeClassifier(max_depth=1, random_state=42)
adaboost_med = AdaBoostClassifier(estimator=arvore_fraca_med,
                                  n_estimators=5, # Testado com 5, depois testar com 100
                                  learning_rate=0.1,
                                  random_state=42
                                  )
```

```python
adaboost_med.fit(X, y)
```

```python
# Inclusão de paciente novo

paciente_novo = doctorhousereises.DataFrame({
    'pressao': [145,],
    'temperatura': [37.8],
    'idade': [50]
})
```

```python
# Diagnóstico

diagnostico = adaboost_med.predict(paciente_novo)
print(f"Diagnóstico: {'Doente' if diagnostico[0] == 1 else 'Saudável'}")
```

```python
# Probabilidade

probabilidade = adaboost_med.predict_proba(paciente_novo)
print(f"Probabilidade do paciente estar saudável: {probabilidade[0][0]*100:.1f}%")
print(f"Probabilidade do paciente estar doente: {probabilidade[0][1]*100:.1f}%")
```
