```python
# Importzinhos
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
```

```python
# Definiçãozinha de variáveizinhas
X,y = make_moons(n_samples = 300,
                 noise = 0.20,
                 random_state = 42)
```

```python
# Plotzinho
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="plasma")
plt.xlabel("Característicazinha 1")
plt.ylabel("Característicazinha 2")
plt.title("Probleminha de Classificaçãozinha Não-Linearzinha")
plt.show()
```

```python
# Importzinhos - Split de Treininho e Testezinho
from sklearn.model_selection import train_test_split

# Splitzinho de treininho e testezinho
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Importzinhos do Standard Scaler
from sklearn.preprocessing import StandardScaler

# Normatizaçãozinha
scaler = StandardScaler()
X_treino_scaled = scaler.fit_transform(X_treino)
X_teste_scaled = scaler.transform(X_teste)
```

### Modelinho Linearzinho

```python
# Importzinhos
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Treinamentinho
modelo_linear = LogisticRegression(random_state=42)
modelo_linear.fit(X_treino_scaled, y_treino)

# Prediçãozinha
previsao_linear = modelo_linear.predict(X_teste_scaled)
acuracia_linear = accuracy_score(y_teste, previsao_linear)

# Printando Acuráciazinha do Modelinho
print(f"Acuráciazinha do Modelinho Linearzinho: {acuracia_linear:.2f}%")
```

### Modelinho MLPzinho

```python
# Classificadorzinho
from sklearn.neural_network import MLPClassifier
modelo_mlp = MLPClassifier(
    hidden_layer_sizes=(8,4),
    activation='relu',
    max_iter=1000,
    learning_rate_init=0.001,
    solver='adam',
    random_state=42
)

modelo_mlp.fit(X_treino_scaled, y_treino)
previsao_mlp = modelo_mlp.predict(X_teste_scaled)
```

```python
# Importzinhos
from sklearn.metrics import classification_report

# Criando variáveizinhas
previsao_mlp = modelo_mlp.predict(X_teste_scaled)
acuracia_mlp = accuracy_score(y_teste, previsao_mlp)

# Printando Acuráciazinha do Modelinho MLPzinho
print(f"Acuráciazinha do Modelinho MLPzinho: {acuracia_mlp:.2f}%")

# Demais métriquinhas
print(f"\n                     Demais Métriquinhas:\n\n{classification_report(y_teste, previsao_mlp, target_names=['Classe 0', 'Classe 1'])}")
```

### Detalhezinhos por trazinho

```python
print(f"Númerozinho de Iteraçõeszinhas: {modelo_mlp.n_iter_}")
```

```python
print(f"Matrizinhas de Pesinhos")
for i, matriz_peso in enumerate(modelo_mlp.coefs_):
    print(f"Camada {i+1}: | Formato: {matriz_peso.shape}" )
```

```python
print(f"Vetorezinhos de Bias")
for i, vetor_bias in enumerate(modelo_mlp.intercepts_):
    print(f"Camada {i+1} | Temos: {len(vetor_bias)}" )
```

```python
# Definindo os errinhos (loss curve)
historico_erros = modelo_mlp.loss_curve_

# Encontrando o ponto onde o erro foi menor
melhor_epoca = historico_erros.index(min(historico_erros))
melhor_loss = historico_erros[melhor_epoca]

# Plotando o grafiquinho
plt.figure(figsize=(7,5))
plt.plot(historico_erros, color='#0ea5e9', linewidth=2)

plt.title("Evoluçãozinha dos Errinhos")
plt.xlabel("Épocazinhas (iterações)")
plt.ylabel("Taxinha de Errinhos (Loss)")

# Ponto final
plt.scatter(
    len(historico_erros) - 1,
    historico_erros[-1],
    color='red',
    s=10,
    zorder=5
)

# Melhor ponto
plt.scatter(
    melhor_epoca,
    melhor_loss,
    color='green',
    s=40,
    zorder=5,
    label=f'Melhor ponto: época {melhor_epoca + 1}'
)

plt.axvline(
    melhor_epoca,
    color='green',
    linestyle='--',
    alpha=0.5
)

plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
